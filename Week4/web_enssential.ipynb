```python
import requests
```

```python
url = "http://wttr.in"
response = requests.get(url)

request = response.request

#Display requrest message
print(f"{request.method} {request.path_url} HTTP/1.1")
for name, value in request.headers.items():
    print(f"{name}: {value}")

print()
# Display response message
print("Status code:", response.status_code)
print("Response headers:", response.headers)
print("Response body:", response.text)
```

```python
# HEAD: retrieve headers without the response body
head_response = requests.head(f"{url}/get")

print("HEAD status:", head_response.status_code)
print("HEAD headers:", head_response.headers)
print("HEAD body:", head_response.text)  # Usually empty
```

# multi-representations

```python
# Europe PMC, a large database of biomedical and life-science publications
url = "https://www.ebi.ac.uk/europepmc/webservices/rest/search?query=australia"
headers = {
    "Accept": "application/xml"
}
response = requests.get(url, headers=headers)

print("Request Accept:", response.request.headers.get("Accept"))
print("Response Content-Type:", response.headers.get("Content-Type"))
print("Status:", response.status_code)
print("Body:", response.text)
```
