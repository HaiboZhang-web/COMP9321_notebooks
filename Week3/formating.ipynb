## apply a function to the entire dataset

```python
#1. Remove extra whitespace and standardise names
df["first_name"]=df["first_name"].apply(lambda x: x.strip() if isinstance(x, str) else x)
df["first_name"].map(repr)
```

```python
#2. Convert postTestScore from strings to numbers
df["postTestScore"] = df["postTestScore"].apply(pd.to_numeric)
df["postTestScore"].dtype
```

```python
#3. Apply a user-defined function
def classify_score(score):
    if score >= 85:
        return "High Distinction"
    elif score >= 75:
        return "Distinction"
    elif score >= 65:
        return "Credit"
    elif score >= 50:
        return "Pass"
    else:
        return "Fail"                  

df["postTestScore"].apply(classify_score)
```
