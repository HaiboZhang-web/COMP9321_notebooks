```python
import pandas as pd
import numpy as np
```

## wide format vs long format

```python
df = pd.DataFrame({
    "student": ["Alice", "Bob", "Alice", "Bob", "Alice", "Bob"],
    "subject": ["math", "math", "english", "english", "science", "science"],
    "score": [85, 78, 90, 82, 88, 80]
})
```

```python
df
```

```python
wide_df=df.pivot(index="student",columns="subject",values="score").reset_index()

print(wide_df)
```

```python
long_df = wide_df.melt(id_vars="student", var_name="subject",value_name="score")
long_df
```

## pivot table

```python
books = pd.read_csv("books.csv")
books
```

```python
Author_summary = pd.pivot_table(
    books,
    index="Author",
    values="Identifier",
    aggfunc="count"
).rename(columns={"Identifier": "Number of Books"})

print(Author_summary)

```

## groupby()

```python
# Count books by publisher
books.groupby("Publisher").size()
```

```python
# Count books by publication place
books.groupby("Place of Publication").size()
```

```python
# Count books by publisher and publication place
books.groupby(["Publisher", "Place of Publication"]).size() #.reset_index(name="Number of Books")
#pub_place.dtype
```

```python
#Count non-missing values for each publisher
books.groupby("Publisher").agg(
    titles=("Title", "count"),
    authors=("Author", "count"),
    places=("Place of Publication", "count")
)
```

```python
# Find publishers with at least two books
books.groupby("Publisher").filter(lambda group: len(group) >= 1)
```
