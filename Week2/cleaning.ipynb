## detect outliers

```python
# using z-score

z_score = (df["postTestScore"]-df["postTestScore"].mean())/df["postTestScore"].std()

print(z_score, df["postTestScore"].mean(), df["postTestScore"].std())

outliers = df[z_score.abs() > 2]

print(outliers)
```

```python
# using the IQR method
Q1 = df["postTestScore"].quantile(0.25)
Q3 = df["postTestScore"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["postTestScore"] < lower_bound) |
    (df["postTestScore"] > upper_bound)
]

print(outliers)
```

```python
# using boxplot
df.boxplot(column="postTestScore",grid=False)
plt.show()
plt.close()
```
