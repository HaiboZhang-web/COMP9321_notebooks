# extract data from a simple html

```python
# <!doctype html>
# <html>
# 	<head><title> A simple page </title></head>
# 	<body>
# 		<p> Some simple content.</p>
# 	</body>
# </html>

with open("index.html") as fp:
    soup = BeautifulSoup(fp, "html.parser")
#print(soup)

# extract the title
title = soup.title.get_text()
print("Title:", title)

# Extract all visible text from the body
content = soup.body.get_text(" ", strip=True)
print("Content:", content)

# Extract all paragraphs
print("Paragraphs:")
for paragraph in soup.find_all("p"):
    print(paragraph.get_text())

# using find 
title1 = soup.find("title").get_text()
print("Title:", title1)
para = soup.find("p").get_text()
print("para:", para)
```

# analyse a complicated html

```python
from urllib.parse import urljoin

url = "unsw_jobs.html"

with open(url, encoding="utf-8") as fp:
    soup = BeautifulSoup(fp, "html.parser")

jobs =[]
# Find table rows
for row in soup.select("tr"):
    cells = row.find_all(["td", "th"])

    
    # Skip headings or incomplete rows
    if len(cells) < 3:
        continue

    title_link = cells[0].find("a")

    if title_link:
        title = title_link.get_text(" ", strip=True)
        job_url = urljoin(url, title_link.get("href", ""))
    else:
        title = cells[0].get_text(" ", strip=True)
        job_url = ""

    location = cells[1].get_text(" ", strip=True)
    closing_date = cells[2].get_text(" ", strip=True)

    jobs.append({
        "title": title,
        "location": location,
        "closing_date": closing_date,
        "url": job_url,
    })

for job in jobs:
    print("Title:", job["title"])
    print("Location:", job["location"])
    print("Closing date:", job["closing_date"])
    print("URL:", job["url"])
    print("-" * 50)
    
```

# combining BeautifulSoup and Re

```python
import re

html = """
<div class="student-list">
    <p>Student: Alice Brown | ZID: z5123456</p>
    <p>Student: Bob Smith | ZID: z6234567</p>
    <p>Student: Carol Lee | ZID: z7345678</p>
</div>
"""

soup = BeautifulSoup(html, "html.parser")

# Find all paragraphs in the student list
paragraphs = soup.find_all("p")
print(paragraphs)

# UNSW ZID pattern: z followed by seven digits
zid_pattern = r"\bz\d{7}\b"

for paragraph in paragraphs:
    text = paragraph.get_text(strip=True)
    zid = re.search(zid_pattern, text)

    if zid:
        print(zid.group())

text = soup.find("div", class_="student-list").get_text(" ", strip=True)

zids = re.findall(zid_pattern, text)
print(zids)
```
