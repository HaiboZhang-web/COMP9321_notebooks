```python
doc.page_count
```

## Extract images from a pdf

```python
from pathlib import Path
output_dir = Path("extracted_images")
output_dir.mkdir(exist_ok=True)
```

```python
# extract all images from the pdf and name the images in the format: page_{page_number}_image_{image_index}.jpeg

with pymupdf.open("acm_template.pdf") as doc:
    for pagenumber, page in enumerate(doc):
        images = page.get_images() # Finds all images embedded on the current page, full=True returns detailed information about each image.
        #print(images)
        # save all images
        for image_index,image_info in enumerate(images): #image_info is a tuple containing metadata about the image.
            #print(pagenumber, image_index,image_info) 
            xref = image_info[0]
            image_data = doc.extract_image(xref)

            #print(image_data)
            #print()
            
            image_ext = image_data["ext"]
            image_bytes = image_data["image"]

            output_path = (output_dir / f"page_{pagenumber}_image_{image_index}.{image_ext}")
            print(output_path)
            
            output_path.write_bytes(image_bytes)
        
                  
```
