## extract text

```python
from pdfminer.high_level import extract_text
```

```python
text = extract_text("acm_template.pdf")
print(text)
```

## Extract text with layout preserved

```python
from pdfminer.layout import LAParams

pdf_path = "acm_template.pdf"

layout_settings = LAParams(
    char_margin=2.0,
    word_margin=0.1,
    line_margin=0.5,
    line_overlap=0.5,
)

text = extract_text(
    pdf_path,
    laparams=layout_settings
)

print(text)
```

## convert pdf to html

```python
from io import BytesIO
from pdfminer.high_level import extract_pages
from pdfminer.converter import HTMLConverter
from pdfminer.layout import LAParams
from pdfminer.pdfinterp import PDFResourceManager

output_buffer = BytesIO()
resource_manager = PDFResourceManager()

converter = HTMLConverter(
    resource_manager,
    output_buffer,
    codec="utf-8",
    laparams=LAParams()
)

for page_layout in extract_pages("acm_template.pdf"):
    converter.receive_layout(page_layout)

converter.close()

with open("output.html", "wb") as html_file:
    html_file.write(output_buffer.getvalue())

output_buffer.close()
```
