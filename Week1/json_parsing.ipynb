```python
import json
import requests
```

## Download the JSON object

```python
response = requests.get("https://hackerone.com/reports/328486.json")

response.raise_for_status()

report = response.json()

#print(report)
print("Report ID:", report["id"])
print("Title:", report["title"])
```

## Access objects

```python
reporter_username = report["reporter"]["username"]
team_handle = report["team"]["handle"]
team_name = report["team"]["profile"]["name"]

print("Reporter:", reporter_username)
print("Team handle:", team_handle)
print("Team name:", team_name)
```

## Extract selected information and save to a file

```python
report_summary = {
    "id": report.get("id"),
    "title": report.get("title"),
    "state": report.get("state"),
    "substate": report.get("substate"),
    "reporter": report.get("reporter", {}).get("username"),
    "team": report.get("team", {}).get("handle"),
    "public": report.get("public"),
    "number_of_votes": report.get("vote_count", 0),
}

print(json.dumps(report_summary, indent=4))

with open("report_summary.json", "w") as file:
    json.dump(report_summary, file, indent=4)
```

## converts JSON data into Python objects

```python
with open("report_summary.json", "r") as file:
    report = json.load(file)

print(type(report))
```

## Modify and save the loaded object

```python
report["state"] = "Resolved"
report["reviewed"] = True

with open("updated_report.json", "w") as file:
    json.dump(report, file, indent=4)

print(report)
```
