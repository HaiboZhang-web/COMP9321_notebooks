# csv module demonstration

```python
import csv
```

## read and write csv data

```python
students = [
    ["ZID", "Name", "Degree", "WAM"],
    ["8787878", "Emily Chole", "1678", "80.5"],
    ["9601234", "John Smith", "8543", "95.1"],
]

with open("students.csv", "w") as file:
    writer = csv.writer(file)
    #writer.writerows(students)
    writer.writerow(students[0])
    writer.writerow(students[1])
    writer.writerow(students[2])

with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
    
```

## Read and write as dictionaries

```python
with open("students.csv", "r") as file:
    reader = csv.DictReader(file)
    print(reader.fieldnames);

    for student in reader:
        print(student["Name"], student["WAM"])
```

```python
students = [
    {"id": 8787878, "name": "Emily Chole", "mark": 80.5},
    {"id": 9601234, "name": "John Smith", "mark": 90.1}
]

with open("students_dict.csv", "w") as file:
    fieldnames = ["id", "name", "mark"]
    writer = csv.DictWriter(file,fieldnames)

    writer.writeheader()
    writer.writerows(students)
```
