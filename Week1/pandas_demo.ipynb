## series

```python
marks = pd.Series([68,98,76])
print(type(marks))
print(marks)
```

```python
# create a Series with customed labels
stu1 = pd.Series(["8787878", "Emily Chole", "1678", "80.5"], ["ZID", "Name", "Degree", "WAM"])
print(stu1)
print(stu1['Name'])
```

```python
# create a Series from a dictionary
stu2 = pd.Series({"ZID":"8787878", "Name":"Emily Chole", "Degree": 1678, "WAM":80.5})
print(stu2)
print(stu2["Name"])
```

## DataFrame

## create a DataFrame

```python
students={"Name": ["Emily","John","Aisha"],
         "Degree": [1678,8543,2468],
         "WAM": [80.5,95.1,87.0]}

df = pd.DataFrame(students)
print(df)
```
