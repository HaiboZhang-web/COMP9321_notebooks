```python
import pandas as pd
import matplotlib.pyplot as plt
import sklearn
from sklearn.linear_model import LinearRegression

df = pd.DataFrame({
    "month": ["Jan","Feb","Mar","Apr","May"],
    "sales": [100,120,90,150,170],
    "marketing_spend": [20,25,15,30,35]
})

df
```

```python
# What is the average sales?

average_sales = df["sales"].mean()

print("Average sales:", average_sales)
```

```python
# Which month had the highest sales?

max_row = df.loc[df["sales"].idxmax()]

print("Month with highest sales:", max_row["month"])

print("Highest sales value:", max_row["sales"])
```

```python
# Is marketing correlated with sales?

correlation = df["sales"].corr(df["marketing_spend"])

print("Correlation between marketing spend and sales:", correlation)
```

```python
# Plot the sales trend
df.plot(kind="line", x="month", y="sales", marker="o", legend=False)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
X = df[["marketing_spend"]]   # independent variable
y = df["sales"]               # dependent variable

model = LinearRegression()
model.fit(X, y)
```

```python
print("Intercept (a):", model.intercept_)
print("Slope (b):", model.coef_[0])

```

```python
# Plot original data
df.plot(kind="scatter", x="marketing_spend", y="sales")

# Plot regression line
plt.plot(df["marketing_spend"], model.predict(X))

plt.title("Linear Regression: Marketing vs Sales")
plt.xlabel("Marketing Spend")
plt.ylabel("Sales")

plt.show()
```
